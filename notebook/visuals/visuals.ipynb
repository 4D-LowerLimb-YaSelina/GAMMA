```python
from pathlib import Path
import numpy as np

shapes['raw'] = {}
shapes['raw']['scale'] = 1
shapes['raw']['ours'] = Path(f'../../output/faust_raw/bench/texture-transfer/full-coverage-soft-canonization/{pair_name}')
shapes['raw']['zoom-out'] = Path(f'../../utils/fmap/experiment/zoomout/xyz-20260728-171839/bench/texture-transfer-raw/{pair_name}')
```

```python
def h_vec(mesh):
    return np.array([0, mesh.points[:, 1].max() - mesh.points[:, 1].min(), 0])

def lowest_point(mesh):
    return np.array([0, mesh.points[:, 1].min(), 0])
```

```python
import numpy as np
pv.set_jupyter_backend('static')
# pv.set_jupyter_backend('html')

# pl = pv.Plotter(window_size=[1920*2, 1080*2])
pl = pv.Plotter(window_size=[1920*4, 1080*4])
pl.enable_parallel_projection()
z_shift = np.array([0.0, 0.0, 0.0])

render_configs = {
    'ambient': 0.1,          # base color visible even in shadow (0–1)
    'diffuse': 1.,          # main shading from light direction (0–1)
    'specular': 0.6,         # shininess / highlight intensity (0–1)
    'specular_power': 30,    # sharpness of specular highlight (0–128)
    'smooth_shading': True,
}

for point in ['5k', '20k', '40k', 'raw']:
    x_shift = np.array([1.7, 0, 0])
    scaler = shapes[point]['scale'] + 0.3

    src_mesh = pv.read(shapes[point]['ours'] / 'src.obj').scale(scaler * 1.1).translate(z_shift)
    src_mesh = src_mesh.translate(h_vec(src_mesh) / 2)
    src_low = lowest_point(src_mesh)
    pl.add_mesh(
        src_mesh,
        texture=pv.read_texture(shapes[point]['ours'] / 'material_0.png'),
        # texture=pv.read_texture(texture_path),
        **render_configs,
    )

    for method in ['ours', 'ulrssm', 'zoom-out']:
        try:
            tgt_mesh = pv.read(shapes[point][method] / 'tgt.obj').scale(scaler).translate(z_shift).translate(x_shift)
            tgt_low = lowest_point(tgt_mesh)
            pl.add_mesh(
                tgt_mesh.translate(src_low - tgt_low),
                texture=pv.read_texture(shapes[point][method] / 'material_0.png'),
                # texture=pv.read_texture(texture_path),
                **render_configs,
            )
        
        except:
            tgt_mesh = pv.read(shapes[point]['ours'] / 'tgt.obj').scale(scaler).translate(z_shift).translate(x_shift)
            pl.add_mesh(
                tgt_mesh.translate(src_low - tgt_low),
                color='white',
                **render_configs,
            )

        x_shift += np.array([1.6, 0, 0])

    z_shift += np.array([0, 0, -1.1])

pl.camera_position = 'xy'
pl.camera.elevation = 15
pl.camera.azimuth = 15
pl.camera.roll += 4
pl.camera.zoom(2)
pl.show()
```

## Noise level

```python
from glob import glob
import numpy as np
import pyvista as pv
pv.set_jupyter_backend('static')
# pv.set_jupyter_backend('html')
pv.global_theme.transparent_background = True

render_configs = {
    'ambient': 0.1,          # base color visible even in shadow (0–1)
    'diffuse': 1.,          # main shading from light direction (0–1)
    'specular': 0.6,         # shininess / highlight intensity (0–1)
    'specular_power': 30,    # sharpness of specular highlight (0–128)
    'smooth_shading': True,
}

pl = pv.Plotter(window_size=[960*2, 650*2])
pl.enable_parallel_projection()

# tr_reg_000
noised_input_ls = sorted(glob(str('../../output/noised_tr_reg_000/*/noisy_input/*.obj')))
x_shift = np.array([1, 0, 0])

for idx in [0, 3, 6]:
    noise_input = noised_input_ls[idx]
    src_mesh = pv.read(noise_input).translate(x_shift)
    pl.add_mesh(
        src_mesh,
        **render_configs,
    )
    x_shift = x_shift + [0.7, 0, 0]

    mhr_mesh = pv.read(noise_input.replace('noisy_input', 'mhr')).translate(x_shift)
    pl.add_mesh(
        mhr_mesh,
        color='white',
        **render_configs,
    )
    x_shift = x_shift + [0.7, 0, 0]

# tr_reg_001
noised_input_ls = sorted(glob(str('../../output/noised_tr_reg_001/*/noisy_input/*.obj')))
x_shift = np.array([1, -1.3, 0])

for idx in [0, 3, 6]:
    noise_input = noised_input_ls[idx]
    src_mesh = pv.read(noise_input).translate(x_shift)
    pl.add_mesh(
        src_mesh,
        **render_configs,
    )
    x_shift = x_shift + [0.7, 0, 0]

    mhr_mesh = pv.read(noise_input.replace('noisy_input', 'mhr')).translate(x_shift)
    pl.add_mesh(
        mhr_mesh,
        color='white',
        **render_configs,
    )
    x_shift = x_shift + [0.7, 0, 0]

pl.camera_position = 'xy'
pl.camera.zoom(1.7)
pl.show()
```
